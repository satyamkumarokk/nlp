# TASK 1 — WordNet Synsets and **Meanings**

```python
import nltk
import pandas as pd

# Download required NLTK resources
nltk.download('wordnet')
nltk.download('omw-1.4')

# Import WordNet
from nltk.corpus import wordnet as wn
```

Select at least 10 English **words**

```python
words = [
    "car",
    "bank",
    "bat",
    "dog",
    "light",
    "plant",
    "book",
    "teacher",
    "student",
    "computer"
]

print("Selected Words:")
print(words)
```

Display synsets, definitions and **examples**

```python
q1_results = []

for word in words:

    synsets = wn.synsets(word)

    print("\n" + "=" * 70)
    print("WORD:", word)

    # Display selected synsets
    for synset in synsets[:5]:

        print("\nSynset Name :", synset.name())
        print("Definition  :", synset.definition())

        examples = synset.examples()

        if examples:
            print("Example     :", examples[0])
        else:
            print("Example     : No example available")

        q1_results.append({
            "Word": word,
            "Synset": synset.name(),
            "Definition": synset.definition(),
            "Example": examples[0] if examples else "No example available"
        })
```

Create **DataFrame**

```python
q1_df = pd.DataFrame(q1_results)

print("\nQ1 Synset DataFrame:")
display(q1_df)
```

Identify words having multiple **synsets**

```python
print("\nWords Having Multiple Synsets:\n")

multiple_meaning_words = []

for word in words:

    synsets = wn.synsets(word)

    if len(synsets) > 1:
        multiple_meaning_words.append(word)

        print(word, "->", len(synsets), "synsets")
```

Discuss at least 3 multiple-meaning **words**

```python
print("\n" + "=" * 70)
print("MULTIPLE MEANINGS OF SELECTED WORDS")
print("=" * 70)

selected_multiple_words = ["bank", "bat", "light"]

for word in selected_multiple_words:

    print("\nWord:", word)

    synsets = wn.synsets(word)

    for synset in synsets[:4]:

        print("Synset     :", synset.name())
        print("Definition :", synset.definition())
```

# **TASK 2 — Hypernyms and Semantic Hierarchy**

Select at least 10 **nouns**

```python
noun_words = [
    "car",
    "dog",
    "cat",
    "teacher",
    "student",
    "computer",
    "book",
    "apple",
    "chair",
    "hospital"
]

print("Selected Nouns:")
print(noun_words)
```

Select a suitable noun **synset**

```python
noun_synsets = {}

for word in noun_words:

    synsets = wn.synsets(word, pos=wn.NOUN)

    if synsets:
        noun_synsets[word] = synsets[0]


print("\nSelected Synsets:\n")

for word, synset in noun_synsets.items():
    print(word, "->", synset.name())
```

Find direct **hypernyms**

```python
direct_hypernym_results = []

for word, synset in noun_synsets.items():

    hypernyms = synset.hypernyms()

    hypernym_names = [h.name() for h in hypernyms]

    direct_hypernym_results.append({
        "Word": word,
        "Synset": synset.name(),
        "Direct Hypernyms":
            ", ".join(hypernym_names)
            if hypernym_names
            else "None"
    })


direct_hypernym_df = pd.DataFrame(direct_hypernym_results)

print("\nDirect Hypernyms:")
display(direct_hypernym_df)
```

Function to trace hypernym **hierarchy**

```python
def get_hypernym_chain(synset, levels=3):

    chain = []

    current = synset

    for i in range(levels):

        hypernyms = current.hypernyms()

        if not hypernyms:
            break

        current = hypernyms[0]

        chain.append(current.name())

    return chain
```

Trace hierarchy for all 10 **nouns**

```python
hierarchy_results = []

for word, synset in noun_synsets.items():

    chain = get_hypernym_chain(
        synset,
        levels=3
    )

    hierarchy_results.append({
        "Word": word,
        "Synset": synset.name(),
        "Hypernym Chain":
            " -> ".join(chain)
            if chain
            else "No hypernym found"
    })

```

Display final **DataFrame**

```python
hierarchy_df = pd.DataFrame(hierarchy_results)

print("\nHypernym Hierarchy:")
display(hierarchy_df)
```

# **TASK 3 — WordNet Semantic Similarity**

Select at least 8 word **pairs**

```python
word_pairs = [
    ("car", "automobile"),
    ("dog", "cat"),
    ("dog", "animal"),
    ("dog", "computer"),
    ("car", "vehicle"),
    ("apple", "banana"),
    ("teacher", "student"),
    ("computer", "book")
]

print("Selected Word Pairs:\n")

for pair in word_pairs:
    print(pair)
```

Calculate WordNet path **similarity**

```python
similarity_results = []

for word1, word2 in word_pairs:

    synsets1 = wn.synsets(word1, pos=wn.NOUN)
    synsets2 = wn.synsets(word2, pos=wn.NOUN)

    if synsets1 and synsets2:

        synset1 = synsets1[0]
        synset2 = synsets2[0]

        similarity = synset1.path_similarity(synset2)

        similarity_results.append({
            "Word 1": word1,
            "Word 2": word2,
            "Synset 1": synset1.name(),
            "Synset 2": synset2.name(),
            "Path Similarity": similarity
        })
```

**store** results in DataFrame

```python
similarity_df = pd.DataFrame(similarity_results)

print("\nSemantic Similarity Results:")
display(similarity_df)
```

Rank pairs by similarity **score**

```python
ranked_similarity_df = similarity_df.sort_values(
    by="Path Similarity",
    ascending=False
).reset_index(drop=True)

print("\nRanked Similarity Results:")
display(ranked_similarity_df)
```

Explain the **results**

```python
print("\nObservations:")
print()

print("1. car and automobile are highly similar because")
print("   they represent nearly the same concept.")

print()

print("2. dog and cat are similar because both are animals")
print("   and belong to related semantic categories.")

print()

print("3. dog and animal are related because a dog is")
print("   a type of animal.")

print()

print("4. dog and computer are dissimilar because they")
print("   belong to very different semantic categories.")

print()

print("5. Higher path similarity means the two synsets")
print("   are closer in the WordNet semantic hierarchy.")
```

# **TASK 4 — Homonyms and Contextual Meaning**

Create two sentences for each **homonym**

```python
contextual_data = [

    # BANK
    {
        "Word": "bank",
        "Sentence": "I deposited money in the bank.",
        "Intended Meaning": "Financial institution",
        "Contextual Clue": "deposited money"
    },

    {
        "Word": "bank",
        "Sentence": "We sat on the bank of the river.",
        "Intended Meaning": "Side of a river",
        "Contextual Clue": "river"
    },


    # BAT
    {
        "Word": "bat",
        "Sentence": "The bat flew out of the cave.",
        "Intended Meaning": "Flying mammal",
        "Contextual Clue": "flew, cave"
    },

    {
        "Word": "bat",
        "Sentence": "He hit the ball with a bat.",
        "Intended Meaning": "Sports equipment",
        "Contextual Clue": "hit the ball"
    },


    # BARK
    {
        "Word": "bark",
        "Sentence": "The dog's bark was very loud.",
        "Intended Meaning": "Sound made by a dog",
        "Contextual Clue": "dog, loud"
    },

    {
        "Word": "bark",
        "Sentence": "The bark of the tree was rough.",
        "Intended Meaning": "Outer covering of a tree",
        "Contextual Clue": "tree, rough"
    },


    # LIGHT
    {
        "Word": "light",
        "Sentence": "The room needs more light.",
        "Intended Meaning": "Illumination",
        "Contextual Clue": "room"
    },

    {
        "Word": "light",
        "Sentence": "This bag is very light.",
        "Intended Meaning": "Not heavy",
        "Contextual Clue": "bag, weight"
    },


    # MATCH
    {
        "Word": "match",
        "Sentence": "He lit the candle with a match.",
        "Intended Meaning": "Small stick used to make fire",
        "Contextual Clue": "lit the candle"
    },

    {
        "Word": "match",
        "Sentence": "Our team won the football match.",
        "Intended Meaning": "Sports game or contest",
        "Contextual Clue": "football, team"
    }
]


context_df = pd.DataFrame(contextual_data)

print("Contextual Sentences:")
display(context_df)
```

Display WordNet **synsets**

```python
homonyms = [
    "bank",
    "bat",
    "bark",
    "light",
    "match"
]

print("\n" + "=" * 70)
print("WORDNET SYNSETS OF HOMONYMS")
print("=" * 70)


for word in homonyms:

    print("\nWORD:", word)

    synsets = wn.synsets(word)

    for synset in synsets[:5]:

        print("\nSynset     :", synset.name())
        print("Definition :", synset.definition())

        examples = synset.examples()

        if examples:
            print("Example    :", examples[0])
```

 Compare contextual **sentences**

```python
print("\n" + "=" * 70)
print("CONTEXTUAL MEANING ANALYSIS")
print("=" * 70)


for index, row in context_df.iterrows():

    print("\nWord:", row["Word"])
    print("Sentence:", row["Sentence"])
    print("Intended Meaning:", row["Intended Meaning"])
    print("Contextual Clue:", row["Contextual Clue"])
```
