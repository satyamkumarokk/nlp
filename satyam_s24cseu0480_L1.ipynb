```python
from nltk.corpus import brown

# Load Brown Corpus
sentences = brown.sents()[:20]

print("Number of selected sentences:", len(sentences))
```

```python
print("First 20 sentences from Brown Corpus:\n")

for i, sentence in enumerate(sentences, start=1):
    print(i, ".", " ".join(sentence))
```

```python
from nltk.tokenize import sent_tokenize

# Join the Brown Corpus sentences into one text
text = " ".join(" ".join(sentence) for sentence in sentences)

# Sentence tokenization
sentence_tokens = sent_tokenize(text)

print("Number of sentence tokens:", len(sentence_tokens))

print("\nFirst 5 sentence tokens:")
for i, sentence in enumerate(sentence_tokens[:5], start=1):
    print(i, ".", sentence)
```

```python
from nltk.tokenize import word_tokenize

# Word tokenization of the first sentence
words = word_tokenize(sentence_tokens[0])

print("Original sentence:")
print(sentence_tokens[0])

print("\nWord tokens:")
print(words)
```

```python
from nltk.corpus import stopwords
import string

# Load English stopwords
stop_words = set(stopwords.words('english'))

cleaned_sentences = []

for sentence in sentence_tokens:

    # Word tokenization
    words = word_tokenize(sentence)

    # Convert to lowercase
    # Keep only alphabetic words
    # Remove stopwords
    cleaned_words = [
        word.lower()
        for word in words
        if word.isalpha() and word.lower() not in stop_words
    ]

    cleaned_sentences.append(cleaned_words)

print("Cleaning completed successfully!")
```

```python
from nltk.corpus import brown

# Get alphabetic words from Brown Corpus
brown_words = [
    word.lower()
    for word in brown.words()
    if word.isalpha()
]

# Required words from the question
required_words = [
    "playing",
    "studies",
    "running",
    "better",
    "cars"
]

# Select 30 Brown Corpus words
selected_words = brown_words[:30]

# Add required words
selected_words.extend(required_words)

# Remove duplicates
selected_words = list(dict.fromkeys(selected_words))

print("Total words:", len(selected_words))
print(selected_words)
```

```python
from nltk.stem import PorterStemmer

# Create Porter Stemmer
stemmer = PorterStemmer()

# Test some words
test_words = [
    "playing",
    "studies",
    "running",
    "better",
    "cars"
]

print("Original -> Stemmed")

for word in test_words:
    print(word, "->", stemmer.stem(word))
```

```python
from nltk.stem import WordNetLemmatizer

# Create lemmatizer
lemmatizer = WordNetLemmatizer()

print("Original -> Lemmatized")

for word in test_words:
    print(word, "->", lemmatizer.lemmatize(word))
```

```python
import pandas as pd

comparison = []

for word in selected_words:

    stemmed = stemmer.stem(word)
    lemmatized = lemmatizer.lemmatize(word)

    comparison.append([
        word,
        stemmed,
        lemmatized
    ])

# Create DataFrame
df = pd.DataFrame(
    comparison,
    columns=[
        "Original Word",
        "Stemmed Word",
        "Lemmatized Word"
    ]
)

display(df)
```

```python
from nltk.corpus import words

# Load English vocabulary
word_list = set(
    word.lower()
    for word in words.words()
    if word.isalpha()
)

print("Number of vocabulary words:", len(word_list))
```

```python
misspelled_words = [
    "recieve",
    "becuase",
    "enviroment",
    "seperate",
    "definately",
    "occured",
    "langauge",
    "programing",
    "studing",
    "teh"
]

print("Misspelled words:")
for word in misspelled_words:
    print(word)
```

```python
from difflib import get_close_matches

def correct_word(word):

    suggestions = get_close_matches(
        word.lower(),
        word_list,
        n=1,
        cutoff=0.6
    )

    if suggestions:
        return suggestions[0]

    return "No suggestion"
```

```python
results = []

for word in misspelled_words:

    suggestion = correct_word(word)

    results.append([
        word,
        suggestion
    ])

spelling_df = pd.DataFrame(
    results,
    columns=[
        "Misspelled Word",
        "Suggested Correction"
    ]
)

display(spelling_df)
```

```python
example_sentences = [
    "I recieve many messages every day.",
    "Python is a powerfull programing langauge.",
    "We should protect the enviroment.",
    "She is definately studing NLP.",
    "The error occured becuase of a seperate problem."
]

for sentence in example_sentences:

    print("\nOriginal:")
    print(sentence)

    words_in_sentence = word_tokenize(sentence)

    corrected_words = []

    for word in words_in_sentence:

        if word.isalpha():
            corrected_words.append(correct_word(word))
        else:
            corrected_words.append(word)

    corrected_sentence = " ".join(corrected_words)

    # Fix spaces before punctuation
    corrected_sentence = corrected_sentence.replace(" ,", ",")
    corrected_sentence = corrected_sentence.replace(" .", ".")
    corrected_sentence = corrected_sentence.replace(" !", "!")
    corrected_sentence = corrected_sentence.replace(" ?", "?")

    print("Corrected:")
    print(corrected_sentence)
```

# Q4 – Unigram, Bigram and Trigram Analysis

```python
from nltk.corpus import brown

# Select first 10,000 words from Brown Corpus
subset_words = brown.words()[:10000]

# Keep only alphabetic words and convert to lowercase
clean_words = [
    word.lower()
    for word in subset_words
    if word.isalpha()
]

print("Number of cleaned words:", len(clean_words))

print("\nFirst 50 words:")
print(clean_words[:50])
```

```python
from nltk.util import ngrams

# Generate unigrams
unigrams = list(ngrams(clean_words, 1))

# Generate bigrams
bigrams = list(ngrams(clean_words, 2))

# Generate trigrams
trigrams = list(ngrams(clean_words, 3))

print("First 10 Unigrams:")
print(unigrams[:10])

print("\nFirst 10 Bigrams:")
print(bigrams[:10])

print("\nFirst 10 Trigrams:")
print(trigrams[:10])
```

```python
from collections import Counter

# Calculate frequencies
unigram_freq = Counter(unigrams)
bigram_freq = Counter(bigrams)
trigram_freq = Counter(trigrams)

print("N-gram frequencies calculated successfully!")
```

```python
print("TOP 10 UNIGRAMS")
print("=" * 40)

for gram, frequency in unigram_freq.most_common(10):

    print(gram, ":", frequency)
```

```python
print("TOP 10 BIGRAMS")
print("=" * 40)

for gram, frequency in bigram_freq.most_common(10):

    print(gram, ":", frequency)
```

```python
print("TOP 10 TRIGRAMS")
print("=" * 40)

for gram, frequency in trigram_freq.most_common(10):

    print(gram, ":", frequency)
```

```python
# Top 10 Unigrams
top_unigrams = pd.DataFrame(
    unigram_freq.most_common(10),
    columns=["Unigram", "Frequency"]
)

# Top 10 Bigrams
top_bigrams = pd.DataFrame(
    [
        (" ".join(gram), frequency)
        for gram, frequency in bigram_freq.most_common(10)
    ],
    columns=["Bigram", "Frequency"]
)

# Top 10 Trigrams
top_trigrams = pd.DataFrame(
    [
        (" ".join(gram), frequency)
        for gram, frequency in trigram_freq.most_common(10)
    ],
    columns=["Trigram", "Frequency"]
)

print("TOP 10 UNIGRAMS")
display(top_unigrams)

print("TOP 10 BIGRAMS")
display(top_bigrams)

print("TOP 10 TRIGRAMS")
display(top_trigrams)
```
